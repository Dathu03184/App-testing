```python
import re

skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]

resumes = [
    """
    Name: Arun Kumar
    Email: arun@gmail.com
    Mobile: 9876543210
    Skills: Python, Java, SQL, Machine Learning
    Experience: 3 years
    """,

    """
    Name: Priya Sharma
    Email: priya@gmail.com
    Mobile: 8765432109
    Skills: Java, SQL, NLP
    Experience: 1 year
    """,

    """
    Name: Rahul Kumar
    Email: rahul@gmail.com
    Mobile: 9123456780
    Skills: Python, NLP, Machine Learning
    Experience: 4 years
    """
]


def extract_information(resume):

    name = re.search(
        r"Name\s*:\s*([A-Za-z .'-]+)",
        resume,
        re.IGNORECASE
    )

    email = re.search(
        r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",
        resume
    )

    mobile = re.search(
        r"(?:\+91[-\s]?)?([6-9]\d{9})",
        resume
    )

    experience = re.search(
        r"(\d+(?:\.\d+)?)\s*(?:years?|yrs?)",
        resume,
        re.IGNORECASE
    )

    found_skills = []

    for skill in skills_list:
        if re.search(
            r"\b" + re.escape(skill) + r"\b",
            resume,
            re.IGNORECASE
        ):
            found_skills.append(skill)

    return {
        "name": name.group(1).strip() if name else "Not Found",
        "email": email.group(0) if email else "Not Found",
        "mobile": mobile.group(1) if mobile else "Not Found",
        "skills": found_skills,
        "experience": float(experience.group(1)) if experience else 0
    }


print("=" * 60)
print("RESUME INFORMATION EXTRACTION SYSTEM")
print("=" * 60)

eligible_candidates = []

for resume in resumes:

    candidate = extract_information(resume)

    print("\nCandidate Name :", candidate["name"])
    print("Email          :", candidate["email"])
    print("Mobile         :", candidate["mobile"])
    print("Skills         :", ", ".join(candidate["skills"]))
    print("Experience     :", candidate["experience"], "years")

    print("Profile Summary:")
    print(
        candidate["name"],
        "has",
        candidate["experience"],
        "years of experience and skills in",
        ", ".join(candidate["skills"])
    )

    if candidate["experience"] >= 2 and "Python" in candidate["skills"]:
        print("Eligibility    : ELIGIBLE")
        eligible_candidates.append(candidate["name"])
    else:
        print("Eligibility    : NOT ELIGIBLE")


print("\n" + "=" * 60)
print("ELIGIBLE CANDIDATES")
print("=" * 60)

for name in eligible_candidates:
    print(name)
```
